### ***ANALYSE TWEETS FROM TWITTER WITH HASHTAGS RELATED TO HEALTHCARE IN THE UK***

```python
# Load the datset and sense check it.

tw = pd.read_csv('tweets.csv')
print(tw.head(), tw.tail(), tw.shape, tw.dtypes, tw.columns)
```

```python
# Determine whether there are missing values.

tw_na = tw[tw.isna().any(axis=1)]
tw_na.shape

tw['tweet_entities_hashtags'].fillna('Unknown', inplace=True)
tw_na.shape
```

```python
# Determine the metadata and the descriptive statistics of the dataset.

print(tw.info, tw.describe())
```

```python
# Determine the frequency  of the count of times a tweet got retweeted and fovorited.

re_tweet = tw['tweet_retweet_count'].value_counts()
print(f"Frequency count of the count of times a tweet got retweeted : {re_tweet}")

favorite_tweet = tw['tweet_favorite_count']
print(f"Frequency  of the count of times a tweet got favorited : {favorite_tweet}")
```

```python
# Create a new dataframe containing only the text.

tweets_text = tw.select_dtypes('object').copy()
tweets_text.head()
```

```python
# Loop through the messages, and create a list of values containing the # symbol.

tags = []
for y in [x.split(' ') for x in tweets_text['tweet_full_text'].values]:
    for z in y:
        if '#' in z:
            # Change to lowercase.
            tags.append(z.lower())
            
```

```python
# Create a pandas series to count the values in the list.

tags_series = pd.Series(tags)
count_tags = tags_series.value_counts()
count_tags
```

```python
# Convert the series to a dataframe.

count_hashtags = count_tags.to_frame().reset_index()
count_hashtags.rename(columns={'index' : 'word'},inplace=True)
print(count_hashtags.columns)
count_hashtags.dtypes
```

```python
# Display records where the count is larger than 10.

data_count = count_hashtags.loc[count_hashtags['count'] > 10]
print(f"List of words containing # with a length greater than 10 : {data_count}")
```

```python
# Remove overrepresented hashtags.

Q1 = data_count['count'].quantile(0.25)
Q3 = data_count['count'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

data_cleaned = data_count[(data_count['count'] >= lower_bound) & (data_count['count'] <= upper_bound)]
```

```python
# Create a barplot 

sns.set_style('white')
sns.set(rc={'figure.figsize':(15, 12)})
pd.options.display.max_colwidth = 200
fig, ax = plt.subplots()
sns.barplot(x = 'word', y = 'count', data = data_cleaned)
plt.title('Count of words having hashtags', fontsize = 16, fontweight = 'bold')
x_labels = data_cleaned['word']
ax.set_xlabel('word', fontsize = 13, fontweight = 'bold')
ax.set_ylabel('count', fontsize = 13, fontweight = 'bold')

ax.set_xticklabels(x_labels, rotation=80)
plt.show
```
