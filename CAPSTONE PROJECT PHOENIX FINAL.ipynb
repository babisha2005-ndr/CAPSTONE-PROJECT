```python
# DROP COLUMNS WITH PERCENTAGE GREATER THAN 60.
```

```python
# FILL NUMERICAL VALUES WITH MEDIAN AND CATEGORICAL WITH MODE FOR COLUMNS WITH 5% to 60% MISSING DATA.

moderate_missing = missing_pct_app[(missing_pct_app >=5) & (missing_pct_app <= 60)].index
```

```python
for col in moderate_missing:
    if app[col].dtype in ['float64', 'int64']:
        app[col] = app[col].fillna(app[col].median())
    else:
        app[col] = app[col].fillna(app[col].mode()[0])
```

```python
# DROP ROWS FOR COLUMS WITH LESS THAN 5% MISSING DATA.

low_missing = missing_pct_app[(missing_pct_app > 0) & (missing_pct_app < 5)].index
```

```python
app = app.dropna(subset = low_missing)
```

```python
# CREATE A LIST FOR HIGH TICKET LOANS.

target_loans = ['Car loan','Loan for the purchace of equipment','Loan for business development', 'Loan for working capital replenishment', 'Mortgage', 'real estate loan']
```

```python
# CREATE A NEW DATAFRAME TO CREATE NEW COLUMNS USING GROUPBY AND LAMBDA FUNCTIONS.

br_new_df = cr_br.groupby('SK_ID_CURR').apply(lambda g: pd.Series({'NUMBER_OF_ACTIVE_ACCOUNTS': g.loc[g['CREDIT_ACTIVE'] == 'Active', 'CREDIT_ACTIVE'].count(),\
                                                                  'NUMBER_OF_CLOSED_ACCOUNTS': g.loc[g['CREDIT_ACTIVE'] == 'Closed', 'CREDIT_ACTIVE'].count(),\
                                                                  'NUMBER_OF_SOLD_ACCOUNTS': g.loc[g['CREDIT_ACTIVE'] == 'Sold', 'CREDIT_ACTIVE'].count(),\
                                                                   'NUMBER_OF_CARLOANS_LAST3M': g.loc[(g['CREDIT_TYPE'] == 'Car loan') & (g['DAYS_CREDIT'] <91), 'DAYS_CREDIT'].count(),\
                                                                   'NO_OF_TYM_CRED_PROL': g['CNT_CREDIT_PROLONG'].sum(),\
                                                                   'MAX_CREDIT_SUM': g['AMT_CREDIT_SUM'].max(),\
                                                                  'MIN_CREDIT_SUM': g['AMT_CREDIT_SUM'].min(),\
                                                                  'DISTINCT_CURR_COUNT': g['CREDIT_CURRENCY'].nunique(),\
                                                                  'ACC_OPENED_LAST_3M': g.loc[g['DAYS_CREDIT'] <91, 'DAYS_CREDIT'].count(),\
                                                                  'ACC_OPENED_LAST_12M': g.loc[g['DAYS_CREDIT'] <365, 'DAYS_CREDIT'].count(),\
                                                                  'COUNT_60_PLUS_OVERDUES': g.loc[g['CREDIT_DAY_OVERDUE'] > 60, 'CREDIT_DAY_OVERDUE'].count(),\
                                                                  'COUNT_HIGH_TKT_LOANS': g.loc[g['CREDIT_TYPE'].isin(target_loans), 'CREDIT_TYPE'].count()}), include_groups=False)
```

```python
# SELECT ONLY THE SPECIFIED COLUMNS THAT NEED IMPUTATION.

targt_cols = ['AMT_PAYMENT', 'DAYS_ENTRY_PAYMENT',]
```

```python
# SET UP THE COLUMNTRANSFER
# remainder='passthrough' ensures that unlisted columns are not altered.

ctin = ColumnTransformer(transformers=[('imputer', IterativeImputer(random_state=42), targt_cols)],\
      remainder='passthrough')
```

```python
# FIT AND TRANSFORM THE DATA.

imputed_array = ctin.fit_transform(ins_pmt_new)
```

```python
# CONVERT BACK TO A DATAFRAME WITH CORRECT COLUMN ORDER.

new_column_ordr = targt_cols + [col for col in ins_pmt_new.columns if col not in targt_cols]
```

```python
ins_imputed= pd.DataFrame(imputed_array, columns=new_column_ordr)
```

```python
ins_imputed.head()
```

```python
# BRING THE DECIMALS IN AUNIFORM FORMAT.

ins_imputed['DAYS_LATE'] = (ins_imputed['DAYS_ENTRY_PAYMENT'] - ins_imputed['DAYS_INSTALMENT']).round(2)
```

```python
ins_imputed['AMT_UNDERPAID'] = (ins_imputed['AMT_INSTALMENT'] - ins_imputed['AMT_PAYMENT']).round(2)
```

```python
ins_imputed.head()
```

```python
# CREATE A NEW DATAFRAME TO CREATE NEW COLUMNS.

ins_grp = ins_imputed.groupby('SK_ID_CURR').apply(lambda s: pd.Series({'NO_OF_TYM_PAID_LATE': s['DAYS_LATE'].count(),\
                                                                       'NO_OF_TYM_UNDERPAID': s['AMT_UNDERPAID'].count()}), include_groups=False)
                                                                      
```
